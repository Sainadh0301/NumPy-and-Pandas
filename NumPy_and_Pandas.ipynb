```python
import pandas as pd
marks = pd.Series([95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])
print("Students Marks:")
print(marks)
```

```python
first_rank = marks[0]
print("Mark of 1st Rank Student:", first_rank)

top_three = marks.loc[['Rank1', 'Rank2', 'Rank3']]
print("Marks of Top 3 Ranks:")
print(top_three)

third_rank = marks.iloc[2]
print("Mark of 3rd Rank Student:", third_rank)

above_90 = marks[marks > 90]
print("Ranks with Marks Greater Than 90:")
print(above_90)
```

```python

marks.loc['Rank1'] = 100


marks = marks.drop('Rank5')


cgpa = marks / 10

print("Updated Marks:")
print(marks)

print("\nCGPA:")
print(cgpa)
```

```python
import pandas as pd

transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],

    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture',
        'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Furniture', 'Electronics'
    ],

    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],

    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})

print("Transaction Details:")
print(transactions)
```

```python
print("Complete Transactions DataFrame:")
print(transactions)


print("First 5 Rows:")
print(transactions.head())


print("Last 5 Rows:")
print(transactions.tail())


print("Shape of DataFrame:")
print(transactions.shape)


print("Column Names:")
print(transactions.columns)


print("Data Types:")
print(transactions.dtypes)


print("ProductCategory and Amount Columns:")
print(transactions[['ProductCategory', 'Amount']])


print("Last 3 Columns:")
print(transactions.iloc[:, -3:])


print("North Region Transactions with Amount Greater Than 200:")
filtered_data = transactions[
    (transactions['Region'] == 'North') &
    (transactions['Amount'] > 200)
]
print(filtered_data)


print("Product Category Value Counts:")
print(transactions['ProductCategory'].value_counts())


print("Unique Regions:")
print(transactions['Region'].unique())


print("Mean Transaction Amount by Region:")
region_mean = transactions.groupby('Region')['Amount'].mean()
print(region_mean)
```

```python

transactions.loc[
    transactions['TransactionID'] == 102, 'Amount'
] = 165

print("After Modifying Transaction 102:")
print(transactions)



transactions['Discount'] = transactions['Amount'] * 0.10

print("After Adding Discount Column:")
print(transactions)

transactions = transactions[
    transactions['TransactionID'] != 109
]

print("After Removing TransactionID 109:")
print(transactions)


transactions.drop(columns=['Discount'], inplace=True)

print("Final DataFrame After Deleting Discount Column:")
print(transactions)
```
